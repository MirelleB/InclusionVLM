```python
#huggingface_token = "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"

# 2. Load Model and Processor
model_id = "google/paligemma-3b-pt-224"

processor = PaliGemmaProcessor.from_pretrained(model_id)
model = PaliGemmaForConditionalGeneration.from_pretrained(
    model_id, torch_dtype=torch.bfloat16, device_map="auto"
)

# 3. Apply LoRA (Efficient Fine-Tuning)
lora_config = LoraConfig(
    r=8, lora_alpha=32, target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05, task_type="CAUSAL_LM"
)
model = get_peft_model(model, lora_config)
```

```python
# 4. Data Preprocessing
def collate_fn(examples):
    images = [example["img"].convert("RGB") for example in examples]
    # PaliGemma uses a specific "captioning" promptcaption en
    texts = ["<image> Generar pie de foto " for _ in range(len(examples))]
    labels = [example["wixarika"] for example in examples] # Use first caption

    inputs = processor(
        text=texts, images=images, suffix=labels,
        return_tensors="pt", padding="longest"
    )
    return inputs

```

```python
path_files = '/content/drive/MyDrive/AmericasNLP-2026/vlm_experiments/polaris_checkpoints/'
# 5. Training Arguments
training_args = TrainingArguments(
    output_dir=path_files+"paligemma_vlm_epoch_10",
    per_device_train_batch_size=4,
    num_train_epochs=10,
    save_strategy="epoch",
    logging_steps=1000,
    learning_rate=5e-5,
    remove_unused_columns=False,
    fp16=False, # Use bfloat16 for PaliGemma
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=reloaded_dataset,
    data_collator=collate_fn,
)

# 6. Train and Save
trainer.train()
model.save_pretrained(path_files+"final_paligemma")
processor.save_pretrained(path_files+"final_paligemma")
```
