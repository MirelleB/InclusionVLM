## Continued Pre-Training

```python
import torch
from datasets import load_dataset, concatenate_datasets
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling
)
from peft import LoraConfig, get_peft_model
```

```python
from datasets import load_from_disk, concatenate_datasets, load_dataset

english_subset_size = 500  # Number of English samples to mix in
path = "/content/drive/MyDrive/AmericasNLP-2026/vlm_experiments/LM_vLM/"


# Load your language
local_ds = load_from_disk(path+"wixarika_monolingual")

# Load a small slice of English (Wiki) to prevent catastrophic forgetting
en_ds = load_dataset("wikimedia/wikipedia", "20231101.en", split="train", streaming=True)
en_ds_list = []
for i, entry in enumerate(en_ds):
    if i >= english_subset_size: break
    en_ds_list.append({"text": entry["text"]})

from datasets import Dataset
en_ds_final = Dataset.from_list(en_ds_list)

# Combine and shuffle
combined_ds = concatenate_datasets([local_ds, en_ds_final]).shuffle(seed=42)
```

```python
# 1. SETTINGS
model_id = "google/gemma-2b"
output_dir = path+"gemma-2b-wixarika"
```

```python
from huggingface_hub import login
login(token="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")
# 2. TOKENIZER & DATA MIXING
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token

def tokenize_fn(examples):
    return tokenizer(examples["text"], truncation=True, max_length=512)

tokenized_ds = combined_ds.map(tokenize_fn, batched=True, remove_columns=["text"])
```

```python
# 3. LOAD MODEL (Bfloat16 for stability)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)

# 4. LORA SETUP
lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj", "k_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    task_type="CAUSAL_LM"
)
model = get_peft_model(model, lora_config)

```

```python
# 5. TRAINING
args = TrainingArguments(
    output_dir=output_dir,
    per_device_train_batch_size=16,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    num_train_epochs=3, # Usually 1-3 epochs for CPT
    bf16=True,
    logging_steps=500,
    save_total_limit=2,
    optim="paged_adamw_8bit"
)

trainer = Trainer(
    model=model,
    args=args,
    train_dataset=tokenized_ds,
    data_collator=DataCollatorForLanguageModeling(tokenizer, mlm=False)
)

trainer.train()

# 6. SAVE & MERGE (Crucial for the PaliGemma swap)
print("Merging LoRA weights into base model...")
merged_model = model.merge_and_unload()
merged_model.save_pretrained(f"{output_dir}-merged")
tokenizer.save_pretrained(f"{output_dir}-merged")

print(f"Success! Specialized LM is ready at {output_dir}-merged")
```
